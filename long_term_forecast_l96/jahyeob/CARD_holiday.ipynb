```python
import pandas as pd
import holidays

# 대한민국 공휴일 객체 생성 (2024~2025년)
kr_holidays = holidays.KR(years=[2024, 2025])

# date 컬럼을 datetime으로 변환
df['date'] = pd.to_datetime(df['date'])

# 공휴일 또는 주말(토/일) 여부 컬럼 추가 (공휴일/주말: 1, 평일: 0)
df['is_holiday'] = df['date'].dt.date.apply(
    lambda x: 1 if (x in kr_holidays or pd.Timestamp(x).weekday() >= 5) else 0
)

df.head()
```

```python
import pandas as pd

add_holidays = ['2025-01-26', '2025-01-27', '2025-01-28', '2025-01-29']
add_holidays = [pd.to_datetime(d).date() for d in add_holidays]

df['date'] = pd.to_datetime(df['date'])
if 'is_holiday' not in df.columns:
    df['is_holiday'] = 0

mask = df['date'].dt.date.isin(add_holidays)
print("마스킹된 row 수:", mask.sum())
print("마스킹된 날짜:", df.loc[mask, 'date'].dt.date.unique())

df.loc[mask, 'is_holiday'] = 1
print(df['is_holiday'].value_counts())
```

```python
# 공휴일에 대한 날짜 출력
print(df[df['is_holiday'] == 1]['date'].dt.date.unique())
```

```python
import pandas as pd

# add_holidays를 datetime.date로 변환
add_holidays = ['2025-01-26', '2025-01-27', '2025-01-28', '2025-01-29']
add_holidays = [pd.to_datetime(d).date() for d in add_holidays]

# df['date']가 datetime64[ns] 타입이어야 함
df['date'] = pd.to_datetime(df['date'])

# 날짜만 추출해서 datetime.date로 변환
date_only = df['date'].dt.date

# is_holiday 업데이트
df.loc[date_only.isin(add_holidays), 'is_holiday'] = 1

# 결과 확인
print(df['is_holiday'].value_counts())
```
