```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager, rc
import matplotlib.dates as mdates
import seaborn as sns

# 한글 폰트 설정
font_path = "C:/Windows/Fonts/malgun.ttf"  # Windows의 경우 '맑은 고딕' 폰트 경로
font = font_manager.FontProperties(fname=font_path).get_name()
rc('font', family=font)

# 파일 경로 및 설정
file_path = "c:\\jahyeob\\CARD\\CARD\\long_term_forecast_l96\\dataset\\KT\\live_ppltn_stts\\live_ppltn_stts_250418_250421.csv"
time_column = "ppltn_time"  # 시간 컬럼
region_column = "area_nm"  # 지역 컬럼
target_column = "ppltn_rate20"  # 예측 대상 컬럼
regions = ["이태원 관광특구", "강남 MICE 관광특구", "명동 관광특구", "홍대 관광특구", "성수카페거리", "홍대입구역(2호선)"]  # 분석할 지역 목록

# 데이터 로드
data = pd.read_csv(file_path, encoding="utf-8")  # UTF-8로 읽기

# 시간 컬럼을 datetime 형식으로 변환
data[time_column] = pd.to_datetime(data[time_column])

# 1. 시간대별 ppltn_rate20 변화 시각화
plt.figure(figsize=(12, 8))
for region in regions:
    region_data = data[data[region_column] == region]
    plt.plot(region_data[time_column], region_data[target_column], label=region)
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d %H:%M'))
plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=6))
plt.xticks(rotation=45)
plt.title("지역별 ppltn_rate20 시간 추이")
plt.xlabel("시간")
plt.ylabel("ppltn_rate20")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()

# 2. 지역별 ppltn_rate20 분포 확인 (박스플롯)
plt.figure(figsize=(12, 8))
sns.boxplot(x=region_column, y=target_column, data=data[data[region_column].isin(regions)])
plt.title("지역별 ppltn_rate20 분포")
plt.xlabel("지역")
plt.ylabel("ppltn_rate20")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. 시간대별 ppltn_rate20 평균 변화 (히트맵)
data['hour'] = data[time_column].dt.hour  # 시간대 추출
hourly_mean = data.groupby(['hour', region_column])[target_column].mean().unstack()
plt.figure(figsize=(12, 8))
sns.heatmap(hourly_mean, cmap="YlGnBu", annot=True, fmt=".1f")
plt.title("시간대별 ppltn_rate20 평균 (지역별)")
plt.xlabel("지역")
plt.ylabel("시간대")
plt.tight_layout()
plt.show()

# 4. ppltn_rate20과 다른 변수 간 상관관계 분석
correlation_data = data.select_dtypes(include=['float64', 'int64']).corr()
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_data, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("변수 간 상관관계")
plt.tight_layout()
plt.show()

# 5. 특정 지역의 ppltn_rate20 시간대별 변화 (라인플롯)
for region in regions:
    region_data = data[data[region_column] == region]
    plt.figure(figsize=(12, 8))
    sns.lineplot(x=region_data[time_column], y=region_data[target_column])
    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d %H:%M'))
    plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=6))
    plt.xticks(rotation=45)
    plt.title(f"{region}의 ppltn_rate20 시간 추이")
    plt.xlabel("시간")
    plt.ylabel("ppltn_rate20")
    plt.grid()
    plt.tight_layout()
    plt.show()
```

```python
# 시간대별 모든 연령대 ppltn_rate 변화 시각화
plt.figure(figsize=(12, 8))
age_columns = [col for col in data.columns if 'ppltn_rate' in col]  # ppltn_rate 관련 컬럼 추출
for age_col in age_columns:
    plt.plot(data[time_column], data[age_col], label=age_col)
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d %H:%M'))
plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=6))
plt.xticks(rotation=45)
plt.title("시간대별 모든 연령대 ppltn_rate 변화")
plt.xlabel("시간")
plt.ylabel("ppltn_rate")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
# 지역별 연령대 분포 확인
plt.figure(figsize=(12, 8))
melted_data = data.melt(id_vars=[time_column, region_column], 
                        value_vars=age_columns, 
                        var_name='연령대', 
                        value_name='ppltn_rate')
sns.boxplot(x='연령대', y='ppltn_rate', hue=region_column, data=melted_data)
plt.title("지역별 연령대 ppltn_rate 분포")
plt.xlabel("연령대")
plt.ylabel("ppltn_rate")
plt.xticks(rotation=45)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
# 시간대별 연령대 평균 변화
data['hour'] = data[time_column].dt.hour  # 시간대 추출
hourly_mean = data.groupby('hour')[age_columns].mean()
plt.figure(figsize=(12, 8))
sns.heatmap(hourly_mean, cmap="YlGnBu", annot=True, fmt=".1f")
plt.title("시간대별 연령대 ppltn_rate 평균")
plt.xlabel("연령대")
plt.ylabel("시간대")
plt.tight_layout()
plt.show()
```

```python
# 연령대 간 상관관계 분석
correlation_data = data[age_columns].corr()
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_data, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("연령대 간 상관관계")
plt.tight_layout()
plt.show()
```

```python
# 특정 지역의 연령대별 ppltn_rate 시간대별 변화
for region in regions:
    region_data = data[data[region_column] == region]
    plt.figure(figsize=(12, 8))
    for age_col in age_columns:
        plt.plot(region_data[time_column], region_data[age_col], label=age_col)
    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d %H:%M'))
    plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=6))
    plt.xticks(rotation=45)
    plt.title(f"{region}의 연령대별 ppltn_rate 시간 추이")
    plt.xlabel("시간")
    plt.ylabel("ppltn_rate")
    plt.legend()
    plt.grid()
    plt.tight_layout()
    plt.show()
```

```python
import re
import matplotlib.pyplot as plt

# 로그 파일 경로
log_path = r'c:\jahyeob\CARD\CARD\long_term_forecast_l96\result_long_term_forecast.txt'

# ppltn_rate20(MD) 관련 실험명 패턴 (예시: ppltn_40_20, ppltn_100_20 등)
target_pattern = re.compile(r'long_term_forecast_ppltn_\d+_20_CARD_.*')

mae_list = []
exp_names = []

with open(log_path, encoding='utf-8') as f:
    lines = f.readlines()

for i, line in enumerate(lines):
    if target_pattern.match(line.strip()):
        # 다음 줄에서 mae 값 추출
        for j in range(i+1, min(i+4, len(lines))):
            if 'mae:' in lines[j]:
                m = re.search(r'mae:([0-9\.eE+-]+)', lines[j])
                if m:
                    mae = float(m.group(1))
                    mae_list.append(mae)
                    exp_names.append(line.strip())
                break

# Plot
plt.figure(figsize=(10,5))
plt.plot(mae_list, marker='o')
plt.title('Batch-wise MAE for ppltn_rate20 (CARD)')
plt.xlabel('Batch (실험 순서)')
plt.ylabel('MAE')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import re
import matplotlib.pyplot as plt

log_path = r'c:\jahyeob\CARD\CARD\long_term_forecast_l96\logs\ShortForecasting\CARD_ppltn_96_48_241101_250610_run_250612_ver2.log'

train_loss = []
val_loss = []

with open(log_path, encoding='utf-8') as f:
    for line in f:
        # Epoch: 1, Steps: 961 | Train Loss: 0.2433872 Vali Loss: 0.2307964 Test Loss: 1.8061137
        m = re.search(r'Train Loss: ([\d\.]+) Vali Loss: ([\d\.]+)', line)
        if m:
            train_loss.append(float(m.group(1)))
            val_loss.append(float(m.group(2)))

plt.figure(figsize=(8,5))
plt.plot(train_loss, label='Train Loss')
plt.plot(val_loss, label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.title('CARD ppltn_96_48 Loss Curve')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('loss_plot_CARD_ppltn_96_48_live_ppltn_stts_241101_250610_preprocessed_real.png')
plt.show()
```

```python
import re
import matplotlib.pyplot as plt

# 로그 파일 경로를 지정하세요.
log_path = r'c:\jahyeob\CARD\CARD\long_term_forecast_l96\logs\ShortForecasting\CARD_ppltn_96_48_241101_250610_run_250612_ver2.log'

# test loss만 추출
test_losses = []
with open(log_path, 'r', encoding='utf-8') as f:
    for line in f:
        # 예시: "Epoch: 1, ... Test Loss: 0.1234 ..."
        match = re.search(r'Test Loss:\s*([0-9.]+)', line)
        if match:
            test_losses.append(float(match.group(1)))

# 곡선 그리기
plt.figure(figsize=(7,4))
plt.plot(range(1, len(test_losses)+1), test_losses, marker='o', color='red', label='Test Loss')
plt.xlabel('Epoch')
plt.ylabel('Test Loss')
plt.title('Test Loss Curve')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
